```python
import pandas as pd
from statsmodels.stats.inter_rater import fleiss_kappa

# Loading the data
data = pd.read_csv('FrenchLabels.csv', delimiter=';', usecols=['Arthur', 'Thomas', 'Gwendal'])

# Creating a copy for comparison
original_data = data.copy()

# Converting values to string and removing rows with any NaN
data = data.dropna()

# Saving the removed data after the first NaN cleaning
removed_na_data = original_data[~original_data.index.isin(data.index)]

# Filtering valid values.
# I: Inconclusive
# C: Contrarty
# D: Favorable 
valid_data = data[(data['Arthur'].isin(["I", "C", "F"])) 
                  & (data['Thomas'].isin(["I", "C", "F"])) 
                  & (data['Gwendal'].isin(["I", "C", "F"]))]

# Saving the removed data after value filtering
removed_filtered_data = data[~data.index.isin(valid_data.index)]

# Updating 'data' to continue with the filtered data
data = valid_data

# Printing the removed data
print("Data removed due to NaN:")
print(removed_na_data)
print("\nData removed due to values outside ['I', 'C', 'F']:")
print(removed_filtered_data)

# Preparing the matrix for Fleiss' Kappa calculation
fleiss_matrix = pd.DataFrame(0, index=data.index, columns=['I', 'C', 'F'])

# Filling the matrix with the count of each label by rater
for rater in ['Arthur', 'Thomas', 'Gwendal']:
    for label in ['I', 'C', 'F']:
        fleiss_matrix[label] += (data[rater] == label).astype(int)

# Calculating Fleiss' Kappa
kappa = fleiss_kappa(fleiss_matrix, method='fleiss')

# Printing the result
print("Calculated Kappa:", kappa)
```

```python
# Função para encontrar o label majoritário
def final_label(row):
    labels = [row['Arthur'], row['Thomas'], row['Gwendal']]
    return max(set(labels), key=labels.count)

# Criar a nova coluna 'label'
data['label'] = data.apply(final_label, axis=1)
data.to_csv("FinalLabels.csv", index=False)
```
