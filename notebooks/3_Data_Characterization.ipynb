```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Function to adjust x-axis ticks to show the start of each year and the end of the last year
def adjust_ticks(ax, monthly_data):
    unique_years = monthly_data['month_year'].dt.year.unique()
    tick_positions = []
    tick_labels = []
    for i, year in enumerate(unique_years):
        year_data = monthly_data[monthly_data['month_year'].dt.year == year]
        first_month = year_data['month_year'].iloc[0]
        tick_positions.append(first_month)
        tick_labels.append(first_month.strftime('%m/%Y'))
        if i == len(unique_years) - 1:  # Add the last month of the last year as well
            last_month = year_data['month_year'].iloc[-1]
            tick_positions.append(last_month)
            tick_labels.append(last_month.strftime('%m/%Y'))
    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels, rotation=45, ha='right')

# Convert published_at to datetime
df['published_at'] = pd.to_datetime(df['published_at'])
df['month_year'] = df['published_at'].dt.strftime('%m/%Y')

# Aggregate data monthly
monthly_data = df.groupby('month_year').agg(
    unique_channels=('channel_id', 'nunique'), 
    videos=('video_id', 'nunique'),
    comments=('comment_id', 'size'),
    likes=('like_count', 'sum'),
    replies=('is_reply', 'sum'),
    unique_users=('author', 'nunique'),

).reset_index()

# Sort data by month_year
monthly_data['month_year'] = pd.to_datetime(monthly_data['month_year'], format='%m/%Y')
monthly_data.sort_values('month_year', inplace=True)

# List of metrics and their corresponding labels
metrics = ['unique_channels', 'videos',  'unique_users', 'comments', 'likes', 'replies']
labels = [ '# Channels', '# Videos', '# Authors', '# Comments', '# Likes', '# Replies']

# Create subplots in a grid of 2 rows and 3 columns
fig, axs = plt.subplots(nrows=2, ncols=3, figsize=(12, 5), sharex=True)

# Configure subplots
for ax, metric, label in zip(axs.flatten()[:6], metrics, labels):
    sns.lineplot(data=monthly_data, x='month_year', y=metric, ax=ax, color='black')
    ax.set_ylabel(label)
    ax.set_xlabel('Month/Year')
    ax.tick_params(axis='x', rotation=45)
    adjust_ticks(ax, monthly_data)  # Adjust x-axis ticks

# Hide the last empty subplot if applicable
if len(axs.flatten()) > len(metrics):
    for unused_ax in axs.flatten()[len(metrics):]:
        unused_ax.set_visible(False)
fig.suptitle('French Data', fontsize=16, y=1.05)
plt.tight_layout()
plt.show()

# Convert published_at to datetime
df_vape_all_languages['published_at'] = pd.to_datetime(df_vape_all_languages['published_at'])
df_vape_all_languages['month_year'] = df_vape_all_languages['published_at'].dt.strftime('%m/%Y')

# Aggregate data monthly
monthly_data = df_vape_all_languages.groupby('month_year').agg(
    unique_channels=('channel_id', 'nunique'), 
    videos=('video_id', 'nunique'),
    comments=('comment_id', 'size'),
    likes=('like_count', 'sum'),
    replies=('is_reply', 'sum'),
    unique_users=('author', 'nunique'),

).reset_index()

# Sort data by month_year
monthly_data['month_year'] = pd.to_datetime(monthly_data['month_year'], format='%m/%Y')
monthly_data.sort_values('month_year', inplace=True)

# List of metrics and their corresponding labels
metrics = ['unique_channels', 'videos',  'unique_users', 'comments', 'likes', 'replies']
labels = [ '# Channels', '# Videos', '# Authors', '# Comments', '# Likes', '# Replies']

# Create subplots in a grid of 2 rows and 3 columns
fig, axs = plt.subplots(nrows=2, ncols=3, figsize=(12, 5), sharex=True)

# Configure subplots
for ax, metric, label in zip(axs.flatten()[:6], metrics, labels):
    sns.lineplot(data=monthly_data, x='month_year', y=metric, ax=ax, color='black')
    ax.set_ylabel(label)
    ax.set_xlabel('Month/Year')
    ax.tick_params(axis='x', rotation=45)
    adjust_ticks(ax, monthly_data)  # Adjust x-axis ticks
    
# Hide the last empty subplot if applicable
if len(axs.flatten()) > len(metrics):
    for unused_ax in axs.flatten()[len(metrics):]:
        unused_ax.set_visible(False)
fig.suptitle('Overall Data', fontsize=16, y=1.05)
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from collections import Counter
```

```python
channel_comments = df.groupby('channel_id').size().reset_index(name='num_comments')

# Plot the CDF for the number of comments a channel receives
plt.figure(figsize=(4, 3))
sns.ecdfplot(data=channel_comments, x='num_comments', color='black')
plt.xlabel('# Channel Comments (x)')
plt.ylabel('CDF')
plt.xscale('log')  # Set the x-axis to logarithmic scale
plt.grid(True)
plt.show()

# Analyze the distribution of the number of comments a user generates
user_comments = df.groupby('author').size().reset_index(name='num_comments')

# Plot the CDF for the number of comments a user generates
plt.figure(figsize=(4, 3))
sns.ecdfplot(data=user_comments, x='num_comments', color='black')
plt.xlabel('# User Comments (x)')
plt.ylabel('CDF')
plt.xscale('log')  # Set the x-axis to logarithmic scale
plt.grid(True)
plt.show()

user_likes = df.groupby('author_channel_id').agg(total_likes=('like_count', 'sum'),author=('author', 'first')).reset_index()

# Plot the CDF for the number of likes a user receives
plt.figure(figsize=(4, 3))
sns.ecdfplot(data=user_likes, x='total_likes', color='black')
plt.xlabel('# User likes (x)')
plt.ylabel('CDF')
plt.xscale('log')  # Set the x-axis to logarithmic scale
plt.grid(True)
plt.title('CDF of the likes a user receives')
plt.show()


#Analyze the distribution of the number of replies a user receive
reply = df.groupby('parent_id').size().reset_index(name = 'num_reply')

# Plot the CDF for the number of reply a comment receive
plt.figure(figsize=(4, 3))
sns.ecdfplot(data=reply, x='num_reply', color='black')
plt.xlabel('# Replies by comments (x)')
plt.ylabel('CDF')
plt.xscale('log')  # Set the x-axis to logarithmic scale
plt.grid(True)
plt.title('CDF of the replies a comment generates')
plt.show()
```
